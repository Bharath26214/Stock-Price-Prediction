```python
import os
import pandas as pd
import matplotlib.pyplot as plt

pred_dir = "../data/predictions"
```

```python
def show_prediction(ticker):
    path = os.path.join(pred_dir, f"{ticker}_predictions.csv")

    if not os.path.exists(path):
        print(f"No prediction file found for {ticker}.")
        return None

    df = pd.read_csv(path)
    print(f"\nPredicted prices for {ticker} (next {len(df)} days):\n")
    display(df)

    # Optional: plot predicted trend
    plt.figure(figsize=(8, 4))
    plt.plot(df["Date"], df["Predicted_Close"], marker='o')
    plt.title(f"{ticker} - Next {len(df)} Days Forecast")
    plt.xlabel("Date")
    plt.ylabel("Predicted Close Price")
    plt.xticks(rotation=45)
    plt.grid(True)
    plt.show()

    return df
```

```python
ticker = input("Enter ticker (e.g., AAPL): ").strip().upper()
show_prediction(ticker)
```
