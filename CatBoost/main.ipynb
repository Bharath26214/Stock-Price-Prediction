```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

import matplotlib.dates as mdates
```

```python
pred_df = pd.read_csv('predictions/2M_predictions.csv')
pred_df.head()
```

```python
df = pred_df[pred_df['Ticker'] == 'AAPL']
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

plt.figure(figsize=(10, 6))
plt.plot(df['Date'], df['Predicted_Price'], marker='o', label='Predicted Price')

plt.xlim(df['Date'].iloc[0], df['Date'].iloc[-1])

plt.gca().xaxis.set_major_locator(mdates.DayLocator(interval=3))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%b %d'))

plt.gcf().autofmt_xdate(rotation=45)

plt.title('Predicted Price (2-Month Forecast)')
plt.xlabel('Date')
plt.ylabel('Predicted Price ($)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
